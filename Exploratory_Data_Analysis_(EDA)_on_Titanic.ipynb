```python
import pandas as pd
import numpy as np
```

```python
# 1. LOAD DATASET & BASIC DATAFRAME OPERATIONS

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("=== FIRST 5 ROWS ===")
print(df.head())

print("\n=== DATASET STRUCTURE & TYPES ===")
df.info()

print("\n=== SUMMARY STATISTICS ===")
print(df.describe(include='all'))
```

```python
# 2. HANDLING MISSING VALUES & DUPLICATES

print("\n=== INITIAL DUPLICATE COUNT ===")
print(df.duplicated().sum())

print("\n=== MISSING VALUES PER COLUMN ===")
print(df.isnull().sum())

# Clean duplicates and missing values
df = df.drop_duplicates()
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop(columns=['Cabin'])

print("\n=== MISSING VALUES AFTER CLEANING ===")
print(df.isnull().sum())
```

```python
# 3. GROUPING AND AGGREGATION ANALYSIS

print("\n=== SURVIVAL RATE BY GENDER ===")
print(df.groupby('Sex')['Survived'].agg(['count', 'sum', 'mean']).rename(
    columns={'count': 'Total', 'sum': 'Survived', 'mean': 'Survival Rate'}
))

print("\n=== SURVIVAL RATE BY PASSENGER CLASS ===")
print(df.groupby('Pclass')['Survived'].agg(['count', 'sum', 'mean']).rename(
    columns={'count': 'Total', 'sum': 'Survived', 'mean': 'Survival Rate'}
))

print("\n=== MEAN FARE AND AGE BY CLASS & GENDER ===")
print(df.groupby(['Pclass', 'Sex'])[['Age', 'Fare']].mean())
```
