```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style and custom palette formatting
sns.set_theme(style="whitegrid")
custom_palette = sns.color_palette("viridis")
```

```python
# 1. Load Dataset (Titanic)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Clean Age column for numerical plotting
df['Age'] = df['Age'].fillna(df['Age'].median())

# Create a figure canvas
plt.figure(figsize=(16, 12))
```

```python
# PLOT 1: HISTOGRAM (Age Distribution)

plt.subplot(2, 3, 1)
sns.histplot(df['Age'], kde=True, color='#2b5c8f', bins=20)
plt.title('1. Age Distribution (Histogram)', fontsize=12, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Frequency')
```

```python
# PLOT 2: BAR CHART (Survival Rate by Passenger Class)

plt.subplot(2, 3, 2)
sns.barplot(data=df, x='Pclass', y='Survived', palette="Set2")
plt.title('2. Survival Rate by Class (Bar Chart)', fontsize=12, fontweight='bold')
plt.xlabel('Passenger Class (Pclass)')
plt.ylabel('Survival Rate')
```

```python
# PLOT 3: SCATTER PLOT (Age vs. Fare)

plt.subplot(2, 3, 3)
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', palette='coolwarm', alpha=0.7)
plt.title('3. Age vs. Fare (Scatter Plot)', fontsize=12, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Fare ($)')
```

```python
# PLOT 4: BOX PLOT (Fare Distribution across Passenger Class)

plt.subplot(2, 3, 4)
sns.boxplot(data=df, x='Pclass', y='Fare', palette="Pastel1")
plt.title('4. Fare Distribution by Class (Box Plot)', fontsize=12, fontweight='bold')
plt.xlabel('Passenger Class')
plt.ylabel('Fare ($)')
plt.ylim(0, 300)
```

```python
# PLOT 5 & 6: CORRELATION HEATMAP

plt.subplot(2, 3, (5, 6))
numeric_df = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()

sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='mako', linewidths=0.5)
plt.title('5. Feature Correlation Matrix (Heatmap)', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
